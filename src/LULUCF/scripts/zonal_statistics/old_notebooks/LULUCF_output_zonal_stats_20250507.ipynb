```python
import re

def parse_metadata_from_uri(uri: str):
    """
    Extracts year, chunk, and variable from an S3 URI string.
    """
    uri = uri.values.tolist()
    # This regex captures year, chunk, and variable in your filename format
    pattern = r'(\d{4}_\d{4}).*?__(\d+_-?\d+_\d+_-?\d+)__([a-zA-Z0-9_]+)(?:_pixel_yr)?_\1'
    match = re.search(pattern, uri)
    
    if match:
        interval = match.group(1)
        chunk_id = match.group(2)
        variable = match.group(3)
    else:
        interval, chunk_id, variable = None, None, None
    
    return interval, chunk_id, variable
```

```python
import xarray as xr

def extract_chunk_id(uri):
    """Extract the chunk ID from a URI using regex."""
    import re
    uri = str(uri)
    match = re.search(r"__(\d+_-?\d+_\d+_-?\d+)__", uri)
    return match.group(1) if match else None

def open_and_label(uri):
    """Open a single raster and attach chunk_id as coordinate."""
    chunk_id = extract_chunk_id(uri)
    da = xr.open_dataset(uri, engine="rasterio").squeeze()  # squeeze band dim
    da = da.expand_dims({"chunk_id": [chunk_id]})  # attach chunk_id
    return da

# Open and label all rasters
gross_emis_CO2_labeled_chunks = [open_and_label(uri) for uri in gross_emis_CO2_tile_uri.values]
gross_emis_all_gases_labeled_chunks = [open_and_label(uri) for uri in gross_emis_all_gases_tile_uri.values]
node_labeled_chunks = [open_and_label(uri) for uri in node_tile_uri.values]

# Concatenate across chunk_id
gross_emis_CO2 = xr.concat(gross_emis_CO2_labeled_chunks, dim="chunk_id").persist()
gross_emis_all_gases = xr.concat(gross_emis_all_gases_labeled_chunks, dim="chunk_id").persist()
nodes = xr.concat(node_labeled_chunks, dim="chunk_id").persist()

gross_emis_CO2_only
gross_emis_all_gases
nodes
```

```python
node_codes = np.array([3222111, 2212120, 3222121, 3212122, 2212110, 3212121, 3222210, 2223200, 
                       2221200, 3212222, 2221100, 5220000, 4100000, 2223100,
                       2212220, 3212221, 2211200, 5210000, 2211100, 4220000, 2212210, 2214100, 2214200, 4210000, 2215200], 
                      dtype=np.uint32)
```

```python
gross_emis_CO2__by_node = xarray_reduce(
    gross_emis_CO2_sub.band_data,
    node_data,
    func='sum',
    expected_groups=(node_codes),
    reindex=ReindexStrategy(
        blockwise=False, array_type=ReindexArrayType.SPARSE_COO
    ),
    fill_value=0
    
)

gross_emis_all_gases__by_node = xarray_reduce(
    gross_emis_all_gases_sub.band_data,
    node_data,
    func='sum',
    expected_groups=(node_codes),
    reindex=ReindexStrategy(
        blockwise=False, array_type=ReindexArrayType.SPARSE_COO
    ),
    fill_value=0
    
)

print(gross_emis_CO2__by_node)
print(gross_emis_all_gases__by_node)
```

```python
gross_emis_CO2_result = gross_emis_CO2__by_node.compute()
gross_emis_all_gases_result = gross_emis_all_gases__by_node.compute()
```
