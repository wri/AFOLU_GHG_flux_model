This notebook compares vegetation model v1.0.5 chunk stats against flox zonal stats by variable, year, and tile_id.
The two outputs should match.
I used a notebook because both outputs are so large that they cannot be opened in Excel and there are lots of combinations (356 tiles x 9 year x >20 outputs) to compare.
I did this with Claude session 'Flux statistics comparison: chunk vs. flox'.

Results:
Chunk stats and flox zonal stats matched for almost all tiles but didn't match for a sizable number in Canada. 
I investigated this and the fault is ultimately on the chunk stats side.
When I ran v1.0.5 on 2026-01-30, chunk stats didn't sum a variable if there was a NaN in the chunk.
(I corrected this later, in time for SOC v1.0.1 run.)
So, chunk stats was just dropping emissions for chunks that had NaN in them,
which was happening specifically in Canada for AGC and BGC emissions but not deadwood and litter. 
The reason NaN was happening for emissions was because I was missing an emission factor in my partial emission factor table.
The emission factor was missing only was for wildfire driver (code 5) in ecozone 2003 (N. American Boreal tundra woodland).
So, this happened under very limited circumstances and affects a pretty restricted set of pixels.

Conclusion:
I reran the vegetation model and post-processing steps for the 19 tiles that contain this cont-eco. I compared the chunk stats from the reran tiles to the chunk stats from the global run and observed the expected large increases in AGC and BGC emissions, primarily because chunk stats in the global run omitted emissions from many chunks. I also compared zonal stats from the global run and new run and observed slight increases in AGC and BGC emissions because of the inclusion of emissions from the partial emission factor that was previously missing. 

With this, I conclude this analysis. The veg model is updated, as are the supplemental/derivative outputs. Everything is aligned. I am going to replace the affected zonal stats tiles from the global run on 2026-06-01 with the Canada re-run tiles to get a complete zonal stats parquet set. I am not going to try to replace the old chunk stats with the new ones, though; that doesn't seem worth it. This also means that the zonal stats folders referred to below will no longer work (because I am giving them a new, combined name), so this notebook can't be rerun in the future. But this record preserves it fine.

# Compares chunk stats and zonal stats sums for tile x year x variable combinations
### Per Claude session 'Flux statistics comparison: chunk vs. flox'

```python
chunk_stats_gross = pd.read_parquet(gross_outputs_parquet)
chunk_stats_gross
```

```python
chunk_stats_net = pd.read_parquet(net_outputs_parquet)
chunk_stats_net
```

```python
# Combine gross and net chunk stats, aggregate to 10x10 deg tile level
chunk_stats_combined = pd.concat([chunk_stats_gross, chunk_stats_net], ignore_index=True)

chunk_tile_agg = (
    chunk_stats_combined
    .groupby(['tile_id', 'pattern', 'years'], dropna=False)['sum_value']
    .sum()
    .reset_index()
    .rename(columns={'years': 'year', 'pattern': 'variable', 'sum_value': 'tile_sum'})
)
chunk_tile_agg['year'] = chunk_tile_agg['year'].astype(int)
chunk_tile_agg['variable'] = chunk_tile_agg['variable'].str.replace('_ha_yr', '', regex=False)  # Removes _ha_yr to match zonal stats output
chunk_tile_agg
```

```python
# Iterates through flox zonal stats parquets to consolidate them to key variables and outputs

flox_parquet_files = sorted(glob.glob(f'{zonal_stats_folder}*.parquet'))
print(f"Found {len(flox_parquet_files)} flox parquet files")

tile_aggs = []
for f in tqdm(flox_parquet_files):
# for f in tqdm(flox_parquet_files[0:2]):  # To test a subset of tiles
    # print(f"Grouping {f}")
    df_tile = pd.read_parquet(f)
    agg = (
        df_tile
        .groupby(['tile_id', 'analysis_layer', 'year'], dropna=False)['value']
        .sum()
        .reset_index()
    )
    tile_aggs.append(agg)

# Combines simplified tile dfs, and drops EF and RF because those aren't in chunk stats output
flox_tile_agg = (
    pd.concat(tile_aggs, ignore_index=True)
    .pipe(lambda df: df[~df['analysis_layer'].isin([
        'AGC_emission_factor_CO2_only__fraction',
        'removal_factor__AGC__MgC',
        'carbon_density__non_soil__MgC_ha',
    ])])
    .rename(columns={'analysis_layer': 'variable', 'value': 'flox_sum'})
)
flox_tile_agg['year'] = flox_tile_agg['year'].astype(int)
flox_tile_agg
```

```python
# Merges chunk stats and zonal stats output tables to get a combined table for comparison

comparison = pd.merge(
    chunk_tile_agg,
    flox_tile_agg,
    on=['tile_id', 'variable', 'year'],
    how='outer',
    indicator=True
)
comparison

only_in_chunk = comparison[comparison['_merge'] == 'left_only']
only_in_flox  = comparison[comparison['_merge'] == 'right_only']
print(f"Tile×variable×year only in chunk stats: {len(only_in_chunk)}")
print(f"Tile×variable×year only in flox stats:  {len(only_in_flox)}")

comparison = comparison[comparison['_merge'] == 'both'].drop(columns='_merge')
comparison

comparison['abs_diff'] = comparison['flox_sum'] - comparison['tile_sum']
comparison['pct_diff'] = np.where(
    comparison['tile_sum'] == 0,
    np.where(comparison['flox_sum'] == 0, 0.0, np.inf),
    (comparison['flox_sum'] - comparison['tile_sum']) / comparison['tile_sum'].abs() * 100
)
comparison
```

```python
# Reports all tile x year x variable combination sums that don't match between chunk stats and zonal stats

MISMATCH_THRESHOLD_PCT = 0.1  # 0.1% difference threshold

mismatches = (
    comparison[comparison['pct_diff'].abs() > MISMATCH_THRESHOLD_PCT]
    .sort_values('pct_diff', key=lambda s: s.abs(), ascending=False)
    .reset_index(drop=True)
)

print(f"Total tile×variable×year combinations compared: {len(comparison)}")
print(f"Mismatches (|pct_diff| > {MISMATCH_THRESHOLD_PCT}%):  {len(mismatches)}")
print(f"Matching:                                        {len(comparison) - len(mismatches)}")
display(mismatches)

tile_summary = (
    mismatches
    .groupby('tile_id')
    .agg(
        n_mismatches=('variable', 'count'),
        max_abs_pct_diff=('pct_diff', lambda x: x.abs().max()),
        variables=('variable', lambda x: sorted(x.unique()))
    )
    .sort_values('max_abs_pct_diff', ascending=False)
    .reset_index()
)
print(f"\nTiles with at least one mismatch: {len(tile_summary)}")
display(tile_summary)
```

## Comparing chunk stats and zonal stats from Canada rerun

```python
# Compares chunk stats from global run against chunk stats from Canada re-run
# Per Claude session 'Flux statistics comparison: chunk vs. flox'
# I expect AGC and BGC to differ dramatically because old chunk stats was missing some chunks with important emissions,
# while deadwood and litter should be the same as before because they weren't affected by this missing emission factor. 

# --- Load Canada rerun gross flux chunk stats (new) ---
new_xlsx = (f'/mnt/c/GIS/git/AFOLU_GHG_flux_model/chunk_stats/vegetation_fluxes_1x1_chunk_statistics_20260603_01_03_27__Canada_rerun_KEEP.xlsx')
new_cs = pd.read_excel(new_xlsx, sheet_name='gross_outputs_1x1')
new_cs['sum_value'] = pd.to_numeric(new_cs['sum_value'], errors='coerce')

# --- Load old global run gross flux chunk stats (Jan 2026 parquet) ---
old_cs = pd.read_parquet(gross_outputs_parquet)
old_cs['sum_value'] = pd.to_numeric(old_cs['sum_value'], errors='coerce')

# --- Filter: four pool emission patterns, tiles in the Canada rerun ---
pool_patterns = [
    'gross_emissions__AGC__MgCO2_ha_yr',
    'gross_emissions__BGC__MgCO2_ha_yr',
    'gross_emissions__deadwood_C__MgCO2_ha_yr',
    'gross_emissions__litter_C__MgCO2_ha_yr',
]
canada_tiles = new_cs['tile_id'].unique()

def agg_tile(df):
    return (
        df[df['pattern'].isin(pool_patterns) & df['tile_id'].isin(canada_tiles)]
        .groupby(['tile_id', 'pattern', 'years'], dropna=False)['sum_value']
        .sum()   # NaN chunks treated as 0, consistent with old behaviour
        .reset_index()
    )

new_agg = agg_tile(new_cs).rename(columns={'sum_value': 'new_sum'})
old_agg = agg_tile(old_cs).rename(columns={'sum_value': 'old_sum'})
old_agg['years'] = old_agg['years'].astype(str)
new_agg['years'] = new_agg['years'].astype(str)

comp = old_agg.merge(new_agg, on=['tile_id', 'pattern', 'years'], how='outer')
comp['diff'] = comp['new_sum'] - comp['old_sum']
comp['pct_change'] = np.where(
    comp['old_sum'].abs() < 1,
    np.inf,
    comp['diff'] / comp['old_sum'].abs() * 100
)
comp['pattern_short'] = comp['pattern'].str.replace('gross_emissions__', '').str.replace('__MgCO2_ha_yr', '')
comp = comp.sort_values(['tile_id', 'pattern_short', 'years']).reset_index(drop=True)

# --- Summary: total change by pool across all tiles and years ---
summary = (
    comp.groupby('pattern_short')[['old_sum', 'new_sum', 'diff']]
    .sum()
    .assign(pct_change=lambda d: d['diff'] / d['old_sum'].abs() * 100)
)
print("=== Total across all Canada rerun tiles and years ===")
print(summary.to_string(float_format='{:,.0f}'.format))
print()

# --- Per-tile summary: total diff across all years, by pattern ---
tile_summary = (
    comp.groupby(['tile_id', 'pattern_short'])[['old_sum', 'new_sum', 'diff']]
    .sum()
    .reset_index()
    .sort_values(['pattern_short', 'diff'], ascending=[True, False])
)
print("=== Per-tile totals (all years) ===")
print(tile_summary.to_string(index=False, float_format='{:,.0f}'.format))
print()

# --- Full detail available ---
print(f"Full comparison: {len(comp)} rows (tile × pattern × year)")
print("Access via `comp` dataframe.")
```

```python
# Compares flox zonal stats from global run against flox zonal stats from Canada re-run
# Per Claude session 'Flux statistics comparison: chunk vs. flox'
# I expect AGC and BGC emissions to be somewhat higher in the new zonal stats because they include partial emissions that use the previously omitted emission factor.
# Deadwood and litter emissions should be unchaged. 

old_zs_folder = '/mnt/c/GIS/AFOLU_flux_model/LULUCF/zonal_statistics/vegetation_v1_0_5_standard_global__full_run__20260601/'
new_zs_folder = '/mnt/c/GIS/AFOLU_flux_model/LULUCF/zonal_statistics/vegetation_v1_0_5_standard_global__Canada_rerun__20260603/'

canada_tiles = [
    '60N_060W', '60N_070W', '60N_080W', '60N_090W', '60N_100W',
    '60N_110W', '60N_120W', '60N_130W', '60N_140W', '60N_150W', '60N_160W',
    '70N_100W', '70N_110W', '70N_120W', '70N_130W', '70N_140W', '70N_150W',
    '70N_160W', '70N_170W',
]

pool_layers = [
    'gross_emissions__AGC__MgCO2',
    'gross_emissions__BGC__MgCO2',
    'gross_emissions__deadwood_C__MgCO2',
    'gross_emissions__litter_C__MgCO2',
]

def load_zonal_stats(folder, tiles):
    aggs = []
    files = sorted(glob.glob(f'{folder}*.parquet'))
    matched = [f for f in files if any(t in f for t in tiles)]
    for f in tqdm(matched, desc=folder.split('/')[-2]):
        df = pd.read_parquet(f)
        agg = (
            df[df['analysis_layer'].isin(pool_layers)]
            .groupby(['tile_id', 'analysis_layer'], dropna=False)['value']
            .sum()
            .reset_index()
        )
        aggs.append(agg)
    return pd.concat(aggs, ignore_index=True)

old_zs = load_zonal_stats(old_zs_folder, canada_tiles)
new_zs = load_zonal_stats(new_zs_folder, canada_tiles)

comp = old_zs.merge(new_zs, on=['tile_id', 'analysis_layer'], suffixes=('_old', '_new'))
comp['diff'] = comp['value_new'] - comp['value_old']
comp['pct_change'] = np.where(
    comp['value_old'].abs() < 1,
    np.inf,
    comp['diff'] / comp['value_old'].abs() * 100
)
comp['layer_short'] = comp['analysis_layer'].str.replace('gross_emissions__', '').str.replace('__MgCO2', '')
comp = comp.sort_values(['layer_short', 'tile_id']).reset_index(drop=True)

# --- Summary by pool across all 19 tiles ---
summary = (
    comp.groupby('layer_short')[['value_old', 'value_new', 'diff']]
    .sum()
    .assign(pct_change=lambda d: d['diff'] / d['value_old'].abs() * 100)
)
print("=== Total across all 19 tiles (all years) ===")
print(summary.to_string(float_format='{:,.0f}'.format))
print()

# --- Per-tile by pool ---
print("=== Per-tile totals (all years) ===")
print(comp[['tile_id', 'layer_short', 'value_old', 'value_new', 'diff', 'pct_change']]
      .to_string(index=False, float_format='{:,.0f}'.format))
```
